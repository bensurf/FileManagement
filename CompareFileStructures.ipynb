```python
import os
```

```python
def get_folder_dict(folder_path):
    # Traverses a folder and creates a dictionary
    # Subfolder names are keys with values the dictionary given by the result of this function applied to the subfolder
    # File names are keys with values the size of the file in bytes

    dict = {}
    
    for dirpath, dirnames, filenames in os.walk(folder_path):
        for filename in filenames:
            if filename[0] == ".":
                continue
            file_path = os.path.join(dirpath, filename)
            try:
                file_size_bytes = os.path.getsize(file_path)
                dict[filename] = file_size_bytes
            except:
                print("Alias: "+file_path)
        for dirname in dirnames:
            if dirname[0] == ".":
                continue
            dir_path = os.path.join(dirpath, dirname)
            dict[dirname] = get_folder_dict(dir_path)
        break

    return dict
```
